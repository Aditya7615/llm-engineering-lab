```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate,load_prompt
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage, ChatMessage

load_dotenv()
model = ChatGroq(model="llama-3.3-70b-versatile", temperature=1.3, max_tokens=100)
```

```python
chat_history = [
    SystemMessage(content="You are a helpful assistant."),
]
```

```python
while True:
    user_input = input("You: ")
    chat_history.append(HumanMessage(content=user_input))
    if user_input == "exit":
        break
    response = model.invoke([user_input])
    chat_history.append(AIMessage(content=response.content))
    print("Bot:", response.content)

print("Chat ended.")
print("Chat History:", chat_history)
```

## langchain messages Types :
### 1. HumanMessage :
humna message is used to represent messages sent by human users in a chat or conversational interface. It typically contains the content of the message and may include additional metadata such as timestamps or user identifiers.
### 2. AIMessage
AIMessage is used to represent messages generated by an artificial intelligence (AI) system in a chat or conversational interface. Similar to HumanMessage, it contains the content of the message and may include metadata such as timestamps or identifiers for the AI system.
### 3. SystemMessage
SystemMessage is used to represent messages that are generated by the system itself, rather than by a human user or an AI. These messages often provide context, instructions, or information about the state of the conversation or system.
### 4. ChatMessage
ChatMessage is a more general class that can represent any type of message in a chat or conversational interface. It can be used as a base class for HumanMessage, AIMessage, and SystemMessage, allowing for polymorphism and easier handling of different message types within the same framework.

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

message = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Hello, who won the world series in 2020?")
]
response = model.invoke(message)
message.append(AIMessage(content=response.content))
print(response.content)
```

## Dynamic message Template :
is a way to create chat prompt templates using a list of tuples, where each tuple consists of a role (like 'system' or 'human') and a corresponding message template string. This approach allows for dynamic generation of chat prompts by filling in placeholders within the message templates based on provided variables.

```python
chat_templates = ChatPromptTemplate([
   # SystemMessage(content="You are a helpful {domain} expert."),
   # HumanMessage(content="Explain in simple terms , what is {topic}"), 
   # this is another way to define chat prompt template using PromptTemplate class
   ('system', "You are a helpful {domain} expert."),
    ('human', "Explain in simple terms , what is {topic}")
])
prompt = chat_templates.format_prompt(domain="science", topic="quantum computing")
response = model.invoke(prompt.to_messages())
print(response.content)
```

## Message Placeholders :
are special markers or variables within message templates that indicate where dynamic content should be inserted. These placeholders are typically enclosed in curly braces (e.g., {username}, {date}) and are replaced with actual values when the message is generated.


```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# chat template with message placeholder

chat_template = ChatPromptTemplate([
    ('system','You are a helpful customer support agent'),
    MessagesPlaceholder(variable_name='chat_history'),
    ('human','{query}')
])

# load chat history from previous conversation
chat_history = []

with open("chat_history.txt", "r") as file:
    chat_history = file.readlines()

print("Previous Chat History: \n", chat_history)

# create prompt
prompt = chat_template.invoke({'chat_history':chat_history, 'query':'Where is my refund'})

print(prompt)
```

## Structured Output :
Structured Output refers to the format in which data is organized and presented in a consistent and predictable manner. In the context of chat messages, structured output ensures that the information conveyed in the messages follows a specific format, making it easier to parse, analyze, and utilize the data effectively. This can include using predefined templates, consistent data fields, and standardized formats for dates, numbers, and other types of information.

### 1. LLM those generate Structured Output :
- Some language models (LLMs) are designed to generate structured output, meaning they can produce responses that adhere to a specific format or structure. This is particularly useful in applications where the output needs to be easily parsed or integrated into other systems, such as generating JSON, XML, or other data formats.
### 2. Structured Output Parsers :
- Structured Output Parsers are tools or components that process and interpret structured output generated by language models. These parsers take the structured data and convert it into a format that can be easily used by applications, such as extracting specific fields, validating data formats, or transforming the output into different representations. They help ensure that the structured output is correctly understood and utilized in various contexts.
### 3. Unstructured to Structured Output :
- Unstructured to Structured Output refers to the process of converting unstructured data, such as free-form text or raw information, into a structured format that can be easily analyzed and processed. This often involves using techniques like natural language processing (NLP) to extract relevant information, identify patterns, and organize the data into predefined structures, such as tables, databases, or specific data formats like JSON or XML. This transformation enables better data management, analysis, and integration with other systems.

Nested TypedDict refers to the use of TypedDicts within other TypedDicts to create more complex and hierarchical data structures. This allows for defining dictionaries that contain other dictionaries as values, enabling a more organized representation of related data.

In TypedDict, you can specify which fields are required and which are optional. Required fields must be present in the dictionary, while optional fields may or may not be included. ### 10. Validating TypedDict Data :
Validating TypedDict Data involves checking that the data conforms to the structure and types defined in a TypedDict. ### 12. Performance Considerations with TypedDict :
Performance Considerations with TypedDict involve understanding the impact of using TypedDicts on the performance of your Python applications. While TypedDicts provide type safety and improved code clarity, they may introduce some overhead during runtime due to type checking and validation. It's important to balance the benefits of using TypedDicts with the potential performance implications, especially in performance-critical applications.
### 13. TypedDict vs Dataclasses :
TypedDict and Dataclasses are both used to define structured data in Python, but they serve different purposes. TypedDict is primarily used for defining dictionary-like structures with specific key-value pairs and types, while Dataclasses are used to create classes that encapsulate data and provide additional functionality, such as methods and default values. Dataclasses also support features like immutability and comparison methods, making them more suitable for complex data models, whereas TypedDict is more lightweight and focused on type safety for dictionary-like data.


```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from typing import TypedDict, Annotated, Optional, Literal

load_dotenv()

model = ChatGroq(model="llama-3.3-70b-versatile", temperature=1.3, max_tokens=10000)

# schema
class Review(TypedDict):

    key_themes: Annotated[list[str], "Write down all the key themes discussed in the review in a list"]
    summary: Annotated[str, "A brief summary of the review"]
    sentiment: Annotated[Literal["pos", "neg"], "Return sentiment of the review either negative, positive or neutral"]
    pros: Annotated[Optional[list[str]], "Write down all the pros inside a list"]
    cons: Annotated[Optional[list[str]], "Write down all the cons inside a list"]
    name: Annotated[Optional[str], "Write the name of the reviewer"]
    

structured_model = model.with_structured_output(Review)

result = structured_model.invoke("""I recently upgraded to the Samsung Galaxy S24 Ultra, and I must say, it’s an absolute powerhouse! The Snapdragon 8 Gen 3 processor makes everything lightning fast—whether I’m gaming, multitasking, or editing photos. The 5000mAh battery easily lasts a full day even with heavy use, and the 45W fast charging is a lifesaver.

The S-Pen integration is a great touch for note-taking and quick sketches, though I don't use it often. What really blew me away is the 200MP camera—the night mode is stunning, capturing crisp, vibrant images even in low light. Zooming up to 100x actually works well for distant objects, but anything beyond 30x loses quality.

However, the weight and size make it a bit uncomfortable for one-handed use. Also, Samsung’s One UI still comes with bloatware—why do I need five different Samsung apps for things Google already provides? The $1,300 price tag is also a hard pill to swallow.

Pros:
Insanely powerful processor (great for gaming and productivity)
Stunning 200MP camera with incredible zoom capabilities
Long battery life with fast charging
S-Pen support is unique and useful
                                 
Review by Nitish Singh
""")

print(result['name'])
```

### Pydantic Models for Structured Output : used for validating and parsing structured output generated by language models.
Pydantic is a powerful library in Python that allows you to define data models with type annotations and provides validation and parsing capabilities. When using Pydantic models for structured output, you can define classes that represent the expected structure of the output data, and Pydantic will handle the validation and parsing of the data according to the defined model. This is particularly useful when working with language models that generate structured output, as it ensures that the output adheres to a specific format and allows for easy integration with other systems or applications. Pydantic models can also be used to define complex nested structures, optional fields, and default values, making it a versatile tool for managing structured data in Python applications.

```python
from pydantic import BaseModel, FieldSerializationInfo
from typing import Optional

class student(BaseModel):
    name: str
    age: Optional[int] = None
    email: str

new_student = {'name':'aditi','age':'2', 'email':'aditya'}
new_student = student(**new_student)
# student_info = student(name="Alice", age=30, email="adityagoyal@gmail.com")
print(new_student)
```

```python
from pydantic import BaseModel, EmailStr, Field
from typing import Optional

class student(BaseModel):

    name: str = 'Aditya Goyal'
    age: Optional[int] = None
    email: EmailStr
    cgpa: float = Field(..., gt=0.0, lt=10, description="CGPA must be between 0.0 and 10.0")

new_student = {'age':'2', 'email':'adityagoayl@gmail.com', 'cgpa': 8.5}
new_student = student(**new_student)
# student_info = student(name="Alice", age=30, email="adityagoyal@gmail.com")

student_dict = dict(new_student)
print(student_dict['age'])

student_json = new_student.model_dump_json()
print(student_json)
```
