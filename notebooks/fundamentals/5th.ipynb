# **Ai Agents :**
Ai agents are autonomous entities that use artificial intelligence to perform tasks, make decisions, and interact with their environment. They can be designed to operate in various domains, such as customer service, healthcare, finance, and more. Ai agents can be categorized into different types based on their capabilities and functionalities:
1. **Reactive Agents**: These agents respond to stimuli from their environment without maintaining an internal state. They make decisions based solely on the current input they receive.
2. **Deliberative Agents**: These agents maintain an internal state and use it to make informed decisions. They can plan and reason about their actions based on past experiences and future goals.
3. **Hybrid Agents**: These agents combine reactive and deliberative capabilities, allowing them to respond to immediate stimuli while also considering long-term goals and plans.
4. **Learning Agents**: These agents can learn from their interactions with the environment and improve their performance over time. They use machine learning techniques to adapt their behavior based on feedback and new data.
5. **Multi-Agent Systems**: These systems consist of multiple agents that interact with each other to achieve common goals or compete for resources. They can collaborate, negotiate, and coordinate their actions to achieve better outcomes.
Ai agents can be implemented using various technologies, including natural language processing, computer vision, machine learning, and robotics. They can be deployed in virtual environments, such as chatbots and virtual assistants, or in physical environments, such as autonomous vehicles and robots. The development of Ai agents has the potential to revolutionize many industries by automating tasks, improving decision-making, and enhancing user experiences.

### Creating an Ai Agent using Langchain

```python
import requests
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from langchain.agents import create_agent

# 1. Setup Model
model = ChatGroq(model="llama-3.3-70b-versatile", temperature=0)

# 2. Updated Tool using OpenWeatherMap
@tool
def get_weather_data(city: str) -> str:
    """
    Fetches real-time weather data for a given city using OpenWeatherMap.
    """
    # Get your free key from openweathermap.org
    api_key = "d71852e80771ae90cfe1919d83f50812" 
    url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}&units=metric"
    
    response = requests.get(url,timeout=10)

    return response.json()

# 3. Create Agent (No ReAct hub prompt needed)
agent_executor = create_agent(
    model,
    tools=[get_weather_data],
    # debug=True 
)

# 4. Run
response = agent_executor.invoke({
    "messages": [("user", "What is the weather in Bhopal?")]
})

print(response["messages"][-1].content)
```

# ReAct :
ReAct (Reasoning and Acting) is a framework for building intelligent agents that can reason about their actions and make informed decisions based on their environment. It combines the concepts of reasoning and acting to create agents that can adapt to changing circumstances and achieve their goals effectively.
In the ReAct framework, agents are designed to have two main components: a reasoning module and an acting module. The reasoning module is responsible for processing information, making inferences, and generating plans based on the agent's goals and the current state of the environment. The acting module is responsible for executing the plans generated by the reasoning module and interacting with the environment to achieve the desired outcomes.
ReAct agents can be used in various applications, such as robotics, virtual assistants, and autonomous systems. They can be designed to perform tasks such as navigation, problem-solving, and decision-making in complex environments. The ReAct framework allows for the development of agents that can learn from their experiences and improve their performance over time, making them more effective in achieving their goals.
