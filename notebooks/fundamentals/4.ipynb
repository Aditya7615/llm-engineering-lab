# **RAG** : 
Retrieval-Augmented Generation (RAG) is a framework that combines retrieval-based methods with generative models to enhance the performance of natural language processing tasks. RAG allows for the retrieval of relevant information from a knowledge base or external sources, which can then be used to generate more accurate and informative responses. The RAG framework typically consists of two main components: a retriever and a generator. The retriever is responsible for fetching relevant information based on a given query, while the generator uses the retrieved information to generate a response. RAG can be used in various applications such as question answering, dialogue systems, and information retrieval, where it can help in providing more accurate and contextually relevant responses by leveraging both retrieval and generation capabilities.

### Components of RAG :
1. **Retriever:** The retriever component is responsible for fetching relevant information from a knowledge base or external sources based on a given query. It can utilize various techniques such as keyword matching, semantic search, or vector-based retrieval to identify and retrieve relevant information that can be used for response generation. The retriever can be designed to work with different types of data sources, such as Wikipedia, web search, databases, or knowledge graphs, depending on the specific use case and requirements of the RAG system.
2. **Document Loader:** The document loader component is responsible for loading and preprocessing the documents that will be used as the knowledge base for retrieval. It can handle tasks such as parsing, cleaning, and structuring the documents to ensure that they are in a suitable format for retrieval. The document loader can also perform tasks such as tokenization, stemming, or stop word removal to enhance the retrieval process and improve the relevance of the retrieved information.
3. **Vector Store:** The vector store component is responsible for storing the documents in a vectorized format that can be efficiently searched and retrieved by the retriever. It can utilize techniques such as word embeddings, sentence embeddings, or other vectorization methods to represent the documents in a high-dimensional space. The vector store can also support various indexing and search algorithms to enable fast and accurate retrieval of relevant information based on the query.
4. **Text Splitter:** The text splitter component is responsible for splitting the documents into smaller chunks or passages that can be more easily processed by the retriever and generator. It can utilize techniques such as sentence splitting, paragraph splitting, or other text segmentation methods to break down the documents into manageable pieces. The text splitter can also perform tasks such as removing irrelevant information, filtering out noise, or selecting specific sections of the documents that are most relevant to the query.

### **Steps to create a RAG system :**
1. **Indexing the documents :** The first step in creating a RAG system is to index the documents that will be used as the knowledge base for retrieval. This involves loading the documents, preprocessing them, and storing them in a vectorized format in a vector store. The indexing process may also involve tasks such as tokenization, stemming, or stop word removal to enhance the retrieval process and improve the relevance of the retrieved information.
2. **Setting up the retriever :** The next step is to set up the retriever component of the RAG system. This involves selecting the appropriate retrieval techniques, such as keyword matching, semantic search, or vector-based retrieval, and configuring the retriever to work with the indexed documents. The retriever may also need to be trained or fine-tuned on the specific domain or dataset to improve its performance and relevance in retrieving information.
3. **Augumentation :** The next step is to augment the retrieved information to provide more context and relevance for the generator component of the RAG system. This can involve techniques such as contextual compression, where the retrieved information is compressed to provide a more concise and relevant response, or multi-query retrieval, where multiple queries are used to retrieve a wider range of relevant information. The augmentation process can help in optimizing the retrieval process and improving the overall performance of the RAG system by providing more relevant and concise information for response generation.
4. **Generation :** The final step is to use the retrieved and augmented information to generate a response. This involves using a generative model, such as a language model, to generate a response based on the retrieved information. The generator can be trained or fine-tuned on the specific domain or dataset to improve its performance and relevance in generating responses. The generated response can then be returned to the user or used for further processing in the RAG system.

### Creating currency converter tool :

```python
from langchain_core.tools import InjectedToolArg, tool
import requests
from typing import Annotated

@tool
def get_conversion_factor(base_currency: str, target_currency: str) -> float:
    """
    This function fetches the currency conversion factor between a given base currency and a target currency
    """
    # For demonstration purposes, we'll return a dummy conversion factor.
    # In a real implementation, you would call a currency conversion API here.
    url = 'https://api.fastforex.io/fetch-one'
    params = {
        'from': base_currency,
        'to': target_currency,
        'api_key': '6d024b07d1-a13388f2b0-tco9uh'
    }
    response = requests.get(url, params=params)
    data =  response.json()

    rate = data['result'][target_currency]
    return {"conversion_rate": rate}


@tool
def convert(base_currency_value: int, conversion_rate: Annotated[float, InjectedToolArg]) -> float:
  """
  given a currency conversion rate this function calculates the target currency value from a given base currency value
  """

  return base_currency_value * conversion_rate
```

```python
import json

for tool_call in ai_message.tool_calls:
  # execute the 1st tool and get the value of conversion rate
  if tool_call['name'] == 'get_conversion_factor':
    tool_message1 = get_conversion_factor.invoke(tool_call)
    # fetch this conversion rate
    conversion_rate = json.loads(tool_message1.content)['conversion_rate']
    # append this tool message to messages list
    messages.append(tool_message1)
  # execute the 2nd tool using the conversion rate from tool 1
  if tool_call['name'] == 'convert':
    # fetch the current arg
    tool_call['args']['conversion_rate'] = conversion_rate
    tool_message2 = convert.invoke(tool_call)
    messages.append(tool_message2)

```
