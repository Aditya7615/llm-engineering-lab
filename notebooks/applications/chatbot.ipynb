```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

model = ChatGroq(model="llama-3.3-70b-versatile", temperature=1.3, max_tokens=100)

response = model.invoke("Explain the theory of relativity in simple terms.")
respons = model.invoke("give me 5 indian names")
print(response)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv

load_dotenv()

llm = ChatGoogleGenerativeAI(model = "gemini-2.5-flash" ,temperature=0.5, max_output_tokens=2000)

response = llm.invoke("Explain the theory of relativity in simple terms.")
print(response)
```

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv

load_dotenv()

# 1. Define the LLM
llm = HuggingFaceEndpoint(
    repo_id="google/gemma-2-9b-it", # Note: Ensure the repo name is correct
    task="text-generation"
)

# 2. Explicitly pass the model_id to ChatHuggingFace
# This prevents the StopIteration error by skipping auto-detection
model = ChatHuggingFace(llm=llm)

result = model.invoke("What is the capital of India")

print(result.content)
```
